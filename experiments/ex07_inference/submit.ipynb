```python
import os, sys, glob, json
```

```python
all_objs = []
files = glob.glob('results-challenge/*.json')

counter = 0
for filename in files:
    with open(filename, 'r') as f:
        objs = json.load(f)
        for obj in objs:
            obj['id'] = counter
            counter += 1
        all_objs.extend(objs)
```

```python
print('len(all_objs) = ', len(all_objs))
print('# of JSONs = ', len(files))

with open('results.json', 'w') as f:
    json.dump(all_objs, f)
```
